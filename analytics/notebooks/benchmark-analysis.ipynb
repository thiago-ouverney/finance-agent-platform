# Análise de runtimes e quantizações

Este notebook compara Ollama, llama.cpp e vLLM por modelo, quantização e cenário. Gere primeiro `analytics/data/benchmark_tagged.csv` com os alvos do Makefile.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='whitegrid')
DATASET = Path('../data/benchmark_tagged.csv')
df = pd.read_csv(DATASET)
df.shape, df.head()
```

```python
metrics = [c for c in df.columns if c.endswith('_p50')]
df[['runtime', 'model', 'scenario', 'phase', 'bmc_tags', *metrics]].describe(include='all').T
```

```python
target = 'time_to_first_token_milliseconds_p50'
plot = df.dropna(subset=[target]).copy()
plt.figure(figsize=(12, 6))
sns.barplot(data=plot, x='runtime', y=target, hue='model', errorbar=None)
plt.title('TTFT p50 por runtime e modelo')
plt.ylabel('TTFT p50 (ms)')
plt.show()
```
